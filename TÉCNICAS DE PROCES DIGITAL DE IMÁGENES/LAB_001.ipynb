# **Ejercicio 1: Figura en Espacio de Color RGB**


```python
import py5
import py5_tools
```

```python
def setup():
    py5.size(400, 400)
    py5.color_mode(py5.RGB, 255)
    py5.no_stroke()

    # Dibujo una elipse y la relleno
    py5.fill(51,119,156)
    py5.ellipse(200, 200, 200, 200)

py5.run_sketch()
```

```python
py5_tools.screenshot()
```

```python
img = py5_tools.screenshot()
img.save("figura_rgb.jpg")
print("La imagen se ha guardado como figura_rgb.jpg")
```

```python
py5.exit_sketch()
```

# **Ejercicio 2: Figura en Espacio de Color HSV**

```python
def setup():
    py5.size(400,400)
    py5.color_mode(py5.HSB, 360, 100, 100)
    py5.no_stroke()
    for i in range(360):
        py5.fill(i, 70, 90)
        py5.rect(i, 0, 1, py5.height)

py5.run_sketch()
```

```python
py5_tools.screenshot()
```

```python
py5.exit_sketch()
```
